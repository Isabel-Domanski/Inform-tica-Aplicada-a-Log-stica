```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Carregar o dataset usando a variável 'path' que está disponível no kernel
csv_files = [f for f in os.listdir(path) if f.endswith('.csv')]
if csv_files:
    csv_path = os.path.join(path, csv_files[0])
    df = pd.read_csv(csv_path)

    # Mostrar os nomes das colunas detectadas para confirmação
    print("Colunas encontradas no dataset:", df.columns.tolist())

    # Identificar as colunas de Temperatura e Faturamento (Revenue)
    # Buscando termos comuns em inglês/português devido à origem do dataset
    col_temp = next((c for c in df.columns if 'temp' in c.lower()), None)
    col_revenue = next((c for c in df.columns if 'revenue' in c.lower() or 'fatur' in c.lower() or 'sales' in c.lower()), None)

    if col_temp and col_revenue:
        plt.figure(figsize=(10, 6))
        sns.scatterplot(data=df, x=col_revenue, y=col_temp, color='red', alpha=0.7)
        plt.title(f'Relação entre Faturamento ({col_revenue}) e Temperatura ({col_temp})')
        plt.xlabel(f'Faturamento ({col_revenue})')
        plt.ylabel(f'Temperatura ({col_temp})')
        plt.grid(True, linestyle='--', alpha=0.5)
        plt.show()
    else:
        print("Não foi possível identificar as colunas de Temperatura ou Faturamento de forma automática.")
else:
    print("Arquivo de dados não encontrado.")
```

```python
# Calcular a correlação de Pearson entre as duas variáveis
correlation = df['Temperature'].corr(df['Revenue'])

print(f"--- Estatísticas Descritivas ---")
display(df.describe())

print(f"\n--- Correlação ---")
print(f"O coeficiente de correlação de Pearson entre Temperatura e Faturamento é: {correlation:.4f}")
if correlation > 0.7:
    print("Indica uma correlação positiva muito forte. Quando a temperatura sobe, o faturamento tende a aumentar significativamente.")
elif correlation > 0.4:
    print("Indica uma correlação positiva moderada.")
else:
    print("Indica uma correlação fraca ou inexistente.")
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Ajustar uma regressão linear (grau 1) usando numpy
# Revenue (y) em função da Temperature (x)
x = df['Temperature']
y = df['Revenue']

slope, intercept = np.polyfit(x, y, 1)

print("--- Equação da Regressão Linear ---")
print(f"Faturamento = {slope:.2f} * Temperatura + {intercept:.2f}")
print(f"\nOnde:")
print(f"- O coeficiente angular (inclinação) é {slope:.2f} (para cada grau a mais na temperatura, o faturamento esperado cresce cerca de ${slope:.2f}).")
print(f"- O coeficiente linear (intercepto) é {intercept:.2f} (faturamento teórico quando a temperatura é zero).")
```

```python
# Visualizar o gráfico com a linha da equação ajustada
plt.figure(figsize=(10, 6))
sns.scatterplot(x=x, y=y, color='blue', alpha=0.6, label='Dados Reais')

# Linha de tendência
x_trend = np.linspace(x.min(), x.max(), 100)
y_trend = slope * x_trend + intercept
plt.plot(x_trend, y_trend, color='red', linewidth=2.5, label=f'Reta: y = {slope:.2f}x + {intercept:.2f}')

plt.title('Regressão Linear: Temperatura vs Faturamento')
plt.xlabel('Temperatura (°C)')
plt.ylabel('Faturamento ($)')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```
