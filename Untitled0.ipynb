```python
import pandas as pd

url = "https://raw.githubusercontent.com/Sainathmudigonda/Student-Result-Analysis/refs/heads/main/student_exam_scores.csv"

df = pd.read_csv(url)

display(df.head())
```

```python
import numpy as np
from sklearn.linear_model import LinearRegression

# Preparar os dados
X = df[['hours_studied']].values
y = df['exam_score'].values

# Ajustar o modelo de regressão linear
model = LinearRegression()
model.fit(X, y)

# Obter o coeficiente angular (slope) e o coeficiente linear (intercept)
a = model.coef_[0]
b = model.intercept_

# Exibir a equação
print(f"Equação da Regressão Linear:")
print(f"exam_score = {a:.4f} * hours_studied + {b:.4f}")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Configurar o estilo
sns.set_theme(style="whitegrid")
plt.figure(figsize=(10, 6))

# Criar o gráfico com linha de regressão
sns.regplot(data=df, x='hours_studied', y='exam_score', scatter_kws={'alpha':0.7}, line_kws={'color':'red'})

# Definir a equação obtida anteriormente
equacao_texto = f"y = {a:.4f} * x + {b:.4f}"

# Adicionar o texto da equação dentro do gráfico
# Posicionando no canto superior esquerdo (coordenadas x=1, y=45 aproximadas)
plt.text(1, 45, equacao_texto, fontsize=12, color='red', weight='bold', bbox=dict(facecolor='white', alpha=0.8))

# Customizações adicionais
plt.title('Dispersão com Linha de Regressão e Equação', fontsize=14)
plt.xlabel('Horas de Estudo', fontsize=12)
plt.ylabel('Nota do Exame', fontsize=12)

plt.show()
```
