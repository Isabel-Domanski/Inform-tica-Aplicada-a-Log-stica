```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import requests
import numpy as np
from scipy import stats

# URL da API do Seade para buscar os dados de nupcialidade
url_api = "https://repositorio.seade.gov.br/api/3/action/package_show?id=seade-nupcialidade"

def plotar_dispersao_com_tendencia(x, y, titulo, x_label, y_label):
    # Remove valores nulos para o cálculo
    mask = ~np.isnan(x) & ~np.isnan(y)
    x_clean = x[mask]
    y_clean = y[mask]

    # Calcula a regressão linear: y = slope * x + intercept
    slope, intercept, r_value, p_value, std_err = stats.linregress(x_clean, y_clean)
    r_quadrado = r_value ** 2

    plt.figure(figsize=(10, 7))

    # Gráfico de dispersão
    sns.scatterplot(x=x_clean, y=y_clean, color='#1f77b4', alpha=0.7, s=80, label='Dados')

    # Linha de tendência
    x_linha = np.linspace(x_clean.min(), x_clean.max(), 100)
    y_linha = slope * x_linha + intercept
    plt.plot(x_linha, y_linha, color='red', linestyle='--', linewidth=2,
             label=f'Tendência: y = {slope:.2f}x + {intercept:.2f}')

    # Configurações visuais
    plt.title(titulo, fontsize=14, pad=15)
    plt.xlabel(x_label, fontsize=11)
    plt.ylabel(y_label, fontsize=11)

    # Texto com as estatísticas no gráfico
    texto_stats = f'Correlação (R): {r_value:.2f}\n$R^2$: {r_quadrado:.2f}\nMédia X (Homens): {x_clean.mean():.1f} anos\nMédia Y (Mulheres): {y_clean.mean():.1f} anos'
    plt.gca().text(0.05, 0.95, texto_stats, transform=plt.gca().transAxes,
                   fontsize=10, verticalalignment='top',
                   bbox=dict(boxstyle='round', facecolor='white', alpha=0.8))

    plt.legend(loc='lower right')
    plt.grid(True, linestyle=':', alpha=0.6)
    plt.tight_layout()
    plt.show()

try:
    # Faz o download dos metadados para encontrar o CSV correto
    response = requests.get(url_api, timeout=10).json()
    resources = response['result']['resources']

    csv_url = None
    for r in resources:
        if r['format'].lower() == 'csv' and 'idade' in r['name'].lower():
            csv_url = r['url']
            break

    if not csv_url:
        for r in resources:
            if r['format'].lower() == 'csv':
                csv_url = r['url']
                break

    if csv_url:
        df = pd.read_csv(csv_url, sep=';', encoding='utf-8')
        col_homens = [c for c in df.columns if 'idade' in c.lower() and ('homem' in c.lower() or 'masc' in c.lower() or 'noivo' in c.lower())]
        col_mulheres = [c for c in df.columns if 'idade' in c.lower() and ('mulher' in c.lower() or 'fem' in c.lower() or 'noiva' in c.lower())]

        if col_homens and col_mulheres:
            x_col = col_homens[0]
            y_col = col_mulheres[0]

            df[x_col] = pd.to_numeric(df[x_col].astype(str).str.replace(',', '.'), errors='coerce')
            df[y_col] = pd.to_numeric(df[y_col].astype(str).str.replace(',', '.'), errors='coerce')

            plotar_dispersao_com_tendencia(
                df[x_col].values,
                df[y_col].values,
                'Idade Média ao Casar: Homens vs Mulheres (Dados Reais)',
                f'Idade Média dos Homens ({x_col})',
                f'Idade Média das Mulheres ({y_col})'
            )
        else:
            print("As colunas de idade média não foram identificadas automaticamente.")
    else:
        print("Não foi possível localizar o arquivo CSV no portal do Seade.")

except Exception as e:
    print(f"Não foi possível obter os dados online (Erro: {e}).")
    print("Plotando o gráfico com a linha de tendência utilizando os dados simulados:")

    np.random.seed(42)
    dados_simulados_x = np.random.normal(32, 2.5, 80)
    dados_simulados_y = dados_simulados_x - np.random.normal(2.5, 0.8, 80)

    plotar_dispersao_com_tendencia(
        dados_simulados_x,
        dados_simulados_y,
        'Idade Média ao Casar (Dados Simulados com Linha de Tendência)',
        'Idade Média dos Homens ao Casar (x)',
        'Idade Média das Mulheres ao Casar (y)'
    )
```
